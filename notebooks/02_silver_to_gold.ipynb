# Construction gold_transactions

### 1. Colonnes Prix m² et surface totale

```python
gold_transactions = dvf_silver.clone()
```

```python
prix_m2_df = (
    gold_transactions
    .group_by(["id_mutation","numero_disposition"], maintain_order=True)
    .agg(pl.col("valeur_fonciere"),pl.sum("surface_reelle_bati").alias("surface_totale"))
              )

prix_m2_df = prix_m2_df.with_columns(
    pl.col("valeur_fonciere").list.unique().list.item().alias("valeur_fonciere")
)

prix_m2_df
```

```python
prix_m2_df = prix_m2_df.with_columns(
    pl.when(pl.col("surface_totale") != 0)
    .then(pl.col("valeur_fonciere")/pl.col("surface_totale"))
    .otherwise(None)
    .alias("prix_m2")
)
```

```python
prix_m2_df.null_count()
```

```python
gold_transactions = gold_transactions.join(prix_m2_df.drop("valeur_fonciere"),
                                           left_on=["id_mutation","numero_disposition"],
                                           right_on=["id_mutation","numero_disposition"],
                                           how="left",
                                           validate="m:1")
```

```python
gold_transactions
```

### 2. Colonnes en lien avec le type de bien

```python
local_df = dvf_silver.group_by(["id_mutation","numero_disposition"], maintain_order=True).agg(pl.col("type_local"))
```

```python
local_df.sample(5)
```

```python
local_df = local_df.with_columns(
    pl.col("type_local").list.len()
    .alias("nb_biens"),
    pl.col("type_local").list.unique()
    .alias("type_biens_unique")
)
local_df
```

```python
local_df = local_df.with_columns(
    pl.when(pl.col("type_biens_unique").list.contains("Appartement"))
    .then(True)
    .otherwise(False)
    .alias("has_appartement"),

    pl.when(pl.col("type_biens_unique").list.contains("Maison"))
    .then(True)
    .otherwise(False)
    .alias("has_maison"),

    pl.when(pl.col("type_biens_unique").list.contains("Dépendance"))
    .then(True)
    .otherwise(False)
    .alias("has_dependance")
)
local_df
```

```python
appart = ["Appartement","Dépendance"]
maison = ["Maison","Dépendance"]
mixte = ["Appartement","Maison","Dépendance"]
```

```python
local_df = local_df.with_columns(
    pl.when(
        (pl.col("type_biens_unique").list.contains("Appartement")) 
        &
        (pl.col("type_biens_unique").list.eval(pl.element().is_in(appart)).list.all())

    )
    .then(pl.lit("Appartement"))

    .when(
        (pl.col("type_biens_unique").list.contains("Maison"))
        &
        (pl.col("type_biens_unique").list.eval(pl.element().is_in(maison)).list.all())
    )
    .then(pl.lit("Maison"))

    .when(
        (pl.col("type_biens_unique").list.contains("Appartement"))
        &
        (pl.col("type_biens_unique").list.contains("Maison"))
        &
        (pl.col("type_biens_unique").list.eval(pl.element().is_in(mixte)).list.all())
    )
    .then(pl.lit("Mixte"))

    .when(
        (pl.col("type_biens_unique").list.contains("Indisponible"))
    )
    .then(pl.lit("Complex"))

    .otherwise(pl.lit("Autre"))
    .alias("type_bien_principal")
)
```

```python
gold_transactions = gold_transactions.join(local_df.drop(["type_local","type_biens_unique"]),
                                           left_on=["id_mutation","numero_disposition"],
                                           right_on=["id_mutation","numero_disposition"],
                                           how="left",
                                           validate="m:1")
```

# Construction gold_location_features

```python
bpe_silver
```

```python
gold_location_features = bpe_silver.clone()
```

```python
gold_location_features = gold_location_features.with_columns(
    pl.when((pl.col("classification_n1") == "Santé et action sociale") 
            &
            (pl.col("classification_n2") != "Total")
            &
            (pl.col("classification_n3") != "Total")
            )
    .then(pl.col("nombre_equipements"))
    .otherwise(pl.lit(0))
    .alias("nb_sante")
    ,

    pl.when((pl.col("classification_n1") == "Enseignement")
            &
            (pl.col("classification_n2") != "Total")
            &
            (pl.col("classification_n3") != "Total")
            )
    .then(pl.col("nombre_equipements"))
    .otherwise(pl.lit(0))
    .alias("nb_enseignement")
    ,

    pl.when((pl.col("classification_n2") == "Services publics")
            &
            (pl.col("classification_n3") != "Total")
            )
    .then(pl.col("nombre_equipements"))
    .otherwise(pl.lit(0))
    .alias("nb_services_publics")
    ,

    pl.when((pl.col("classification_n1") == "Transports et déplacements")
            &
            (pl.col("classification_n2") != "Total")
            &
            (pl.col("classification_n3") != "Total")
            )
    .then(pl.col("nombre_equipements"))
    .otherwise(pl.lit(0))
    .alias("nb_transports")
    ,

    pl.when((pl.col("classification_n1") == "Commerces")
            &
            (pl.col("classification_n2") != "Total")
            &
            (pl.col("classification_n3") != "Total")
            )
    .then(pl.col("nombre_equipements"))
    .otherwise(pl.lit(0))
    .alias("nb_commerces")
    ,

    pl.when((pl.col("classification_n1") == "Sports, loisirs et culture")
            &
            (pl.col("classification_n2") != "Total")
            &
            (pl.col("classification_n3") != "Total")
            )
    .then(pl.col("nombre_equipements"))
    .otherwise(pl.lit(0))
    .alias("nb_sports_loisirs")
)
```

```python
gold_location_features = (
    gold_location_features
    .group_by(["annee","code_postal"])
    .agg(
        pl.sum("nb_sante"),
        pl.sum("nb_enseignement"),
        pl.sum("nb_services_publics"),
        pl.sum("nb_transports"),
        pl.sum("nb_commerces"),
        pl.sum("nb_sports_loisirs")
    )
)
```

```python
equipment_cols = [
    "nb_sante",
    "nb_enseignement",
    "nb_services_publics",
    "nb_transports",
    "nb_commerces",
    "nb_sports_loisirs"
]

weights = {
    "nb_sante": 0.25,
    "nb_enseignement": 0.20,
    "nb_transports": 0.20,
    "nb_commerces": 0.15,
    "nb_sports_loisirs": 0.10,
    "nb_services_publics": 0.10,
}
```

```python
data = {"code_postal": ["75001","75002","75003","75004","75005","75006","75007","75008","75009","75010","75011","75012","75013","75014","75015","75016","75017","75018","75019","75020"],
        "surface_km2": [1.824613, 0.991154, 1.170883, 1.600586, 2.539375, 2.153096, 4.090057, 3.880036, 2.178303, 2.891739, 3.665442, 16.314783, 7.149311, 5.614877, 8.494994, 16.372542, 5.668835, 5.996051, 6.792651, 5.983446]
        }
```

```python
surface_km2 = pl.DataFrame(data)
```

```python
gold_location_features = gold_location_features.join(surface_km2,
                                                     on="code_postal",
                                                     how="left")
```

```python
gold_location_features = (
    gold_location_features
    .with_columns([
        (
            pl.col(col) / pl.col("surface_km2")
        )
        .log1p()
        .alias(f"{col}_densite_log")
        for col in equipment_cols
    ])
    .with_columns([
        (
            (pl.col(f"{col}_densite_log") - pl.col(f"{col}_densite_log").min().over("annee"))
            / 
            (pl.col(f"{col}_densite_log").max().over("annee") - pl.col(f"{col}_densite_log").min().over("annee"))
        )
        .fill_nan(0)
        .alias(f"{col}_score")
        for col in equipment_cols
    ])
    .with_columns(
        (
            100 * sum(
                pl.col(f"{col}_score") * weight
                for col, weight in weights.items()
            )
        )
        .round(2)
        .alias("score_equipements")
    )
    .drop([f"{col}_densite_log" for col in equipment_cols])
    .drop([f"{col}_score" for col in equipment_cols])
)
```
